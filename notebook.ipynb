```python
import pandas as pd
import plotly.express as px
```

```python
def read_spot_prices():
    df = pd.read_csv('./data/master/ocean_solutions_master.csv')
    df['periodType'] = df['periodType'].str.upper()
    df = df[df['periodType']=='SPOT']
    df.drop(columns=['source','periodType'], inplace=True)
    return df

data = read_spot_prices()
print(data)
```

```python
fig = px.line(data, x='date', y='price', color='instrument', title='PMT Spot Prices')
#put axis labels at top of graph
fig.update_layout(xaxis_title='Date', yaxis_title='Price', legend_title='Instrument', legend=dict(orientation='h', yanchor='bottom', y=1.02, xanchor='right', x=1))
fig.show()
```

```python
df = data.copy()
df = df.pivot(index='date', columns='instrument', values='price')
df['TD25-TD20'] = df['TD25'] / df['TD20']
df['TD25-TD3C'] = df['TD25'] / df['TD3C']
df['TD20-TD3C'] = df['TD20'] / df['TD3C']

fig = px.line(df, x=df.index, y=['TD25-TD20', 'TD25-TD3C', 'TD20-TD3C'], title='Price Ratio')
#put axis labels at top of graph
fig.update_layout(xaxis_title='Date', yaxis_title='Price Ratio', legend_title='Price Ratio', legend=dict(orientation='h', yanchor='bottom', y=1.02, xanchor='right', x=1))
fig.show()
```
